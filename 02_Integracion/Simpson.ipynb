```python
import numpy as np

# Parámetros del problema
m0 = 140000
q = 2100
u = 2000
g = 9.81
a = 8    
b = 30
valor_verdadero = 11057.16

def v(t):
    return u * np.log(m0 / (m0 - q * t)) - g * t

#pasos
n = 4
def simpson(f, a, b, n):
    # Calculamos el tamaño de cada subintervalo (h)
    h = (b - a) / n
    suma_impares = 0.0
    suma_pares = 0.0

    # Calculamos la suma para los índices impares (i = 1, 3, 5... n-1)
    for i in range(1, n, 2):
        x_i = a + i * h
        suma_impares += f(x_i)

    # Calculamos la suma para los índices pares (i = 2, 4, 6... n-2)
    for i in range(2, n - 1, 2):
        x_i = a + i * h
        suma_pares += f(x_i)

    integral = (h / 3) * (f(a) + 4 * suma_impares + 2 * suma_pares + f(b))
    
    return integral
valor_aprox = simpson(v, a, b, n)
#Error absoluto:
error_absoluto = np.abs(valor_verdadero - valor_aprox)

#Error relativo:
error_relativo = (error_absoluto / np.abs(valor_verdadero)) * 100

# Resultados
print("a) Estimación con Simpson:", valor_aprox)
print("Valor real de la integral:", valor_verdadero)
print("b) Error Absoluto:", error_absoluto)
print("c) Error Relativo:", error_relativo, "%")
```
