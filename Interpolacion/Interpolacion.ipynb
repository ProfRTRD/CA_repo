```python
import numpy as np
#Interpolacion
def Vandermonde(x_datos):
    n = len(x_datos)
    V = np.zeros((n, n), dtype = float)
    for i in range(n):
        for j in range(n):
            V[i,j] = x_datos[i]**j
    return V

def gauss_solve(A, b):
    n = len(b)
    M = np.column_stack((A, b)).astype(float)

    for k in range (n):
        max_idx = np.argmax(np.abs(M[k:n, k]))
        piv = k + max_idx

        if k != piv:
            M[[k, piv]] = M[[piv, k]]
            
        for i in range(k+1, n):
            factor = M[i,k] / M[k,k]
            for j in range(k, n+1):
                M[i,j] = M[i,j] - factor * M[k,j]
    #sustitucion hacia atras
    x = np.zeros(n, dtype = float)
    for i in range(n-1, -1, -1):
        suma = 0
        for j in range(i+1, n):
            suma = suma + M[i,j] * x[j]
        x[i] = (M[i,n] - suma) / M[i,i]
    return x

def interp_directo(x_datos, y_datos, x_eval):
    V = Vandermonde(x_datos)
    a = gauss_solve(V, y_datos)    #coeficientes del polinomio

    resultado = 0
    for j in range(len(a)):
        resultado = resultado + a[j] * (x_eval**j)
    return resultado

        
        
```

```python
data = np.load('data.npy')

#seleccionamos todas las filas de las respectivas columnas
t_dias = data[:, 0]
flujo_relativo = data[:, 1]

#Definir el punto de evaluación
x_eval = -0.25

# Encontrar los índices de los puntos más cercanos a t = -0.25
distancias = np.abs(t_dias - t_eval)
indices_ordenados = np.argsort(distancias)

#agarramos las N muestras mas cercanas al punto de evaluacion
N_puntos = 2
indices_seleccionados = indices_ordenados[:N_puntos]

# Extraemos los datos locales correspondientes
x_local = t_dias[indices_seleccionados]
y_local = flujo_relativo[indices_seleccionados]

print(f"Tiempos más cercanos seleccionados:\n{x_local}")
print(f"Flujos correspondientes:\n{y_local}")

#Calcular el flujo interpolado
flujo_estimado = interp_directo(x_local, y_local, x_eval)
print(flujo_estimado)
```
