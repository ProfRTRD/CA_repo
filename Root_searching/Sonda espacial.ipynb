# Sonda espacial (o asteroide)



```python
import numpy as np
```

```python
#Estacion 1 (x1, y1) = (0,0) Mm
x1 = 0.0
y1 = 0.0
#Estacion 2 (x2. y2) = (50,0) Mm
x2 = 50.0
y2 = 0.0
#Distancias
d1 = 30.0
d2 = 40.0

#Funciones
#Sistema 1
def system(x, y):
    u = (x - x1)**2 + (y - y1)**2 - d1**2
    v = (x - x2)**2 + (y - y2)**2 - d2**2

    return np.array([[u],[v]])
def jacobian(x, y):
    return np.array([
        [2*(x - x1) , 2*(y - y1)],
        [2*(x - x2) , 2*(y - y2)]
    ])

def inv(A):
    a, b = A[0]
    c, d = A[1]
    det = a*d - b*c
    return (1/det) * np.array([[d, -b],
                                [-c, a]])
def newton_raphson(x0, y0 ,ite):
    xi = np.array([[x0], [y0]]) #valores iniciales
    print("valores iniciales:", xi)

    for i in range(1, ite + 1):
        x, y = xi[0, 0], xi[1, 0] #Indica a el solver en que ubicacion encontrar los valores
        F = system(x, y)
        JK = inv(jacobian(x, y))
        xi = xi - JK @ F

    return xi

raiz = newton_raphson(20, 20, 3)
print(raiz)
```
