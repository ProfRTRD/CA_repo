```python
import numpy as np
```

```python
#Modelado del flujo
#F(t) = 100 + 15*sin((pi*t)/6) + 3*sin((pi*t)/2)
def F(t):
    return 100 + (15 * np.sin((np.pi * t)/6)) + (3 * np.sin((np.pi * t)/2))
def F_prime_exact(t):
    return (5*np.pi/2) * np.cos(np.pi*t/6) + (3*np.pi/2) * np.cos(np.pi*t/2)

#diferencias
#dif fwrd
def diff_forward(func, x, h):
    return (func(x + h) - func(x)) / h

#dif bcwrd
def diff_backward(func, x, h):
    return (func(x) - func(x-h)) / h

#dif centr
def diff_centered(func, x, h):
    return (func(x+h) - func(x-h)) / (2*h)

```

```python
#evaluar en 10, 11, 12, 13 y 14 mins
print("--- Evaluacion de F en t = mins ---")
print("Flujo a los 10 mins:",F(10))
print("Flujo a los 11 mins:",F(11))
print("Flujo a los 12 mins:",F(12))
print("Flujo a los 13 mins:",F(13))
print("Flujo a los 14 mins:",F(14), "\n")
#Evaluacion de F'(12)
#Parametros
h = 1
t = 12
exact = F_prime_exact(t)
fwd = diff_forward(F, t, h)
bwd = diff_backward(F, t, h)
ctr = diff_centered(F, t, h)

err_fwd = abs((exact - fwd) / exact) * 100
err_bwd = abs((exact - bwd) / exact) * 100
err_ctr = abs((exact - ctr) / exact) * 100
print("--- Comparacion de F exacta vs diferenciacion en t = 12 ---")
print("Valor exacto:", exact)
print("Valor forward;", fwd, "| \t", "Error relativo:", err_fwd)
print("Valor bacwards;", bwd, "| \t", "Error relativo:", err_bwd)
print("Valor centered;", ctr, "| \t", "Error relativo:", err_ctr)
```
