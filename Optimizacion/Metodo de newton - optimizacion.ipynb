# Seccion aurea

```python
import numpy as np


def ang(x):
    return 4 * np.sin(x) * (1 + np.cos(x))


xl = 0.0
xu = np.pi / 2.0
eps = 0.05
R = (np.sqrt(5) - 1) / 2

d = R * (xu - xl)
x1 = xl + d
x2 = xu - d


iteracion = 1

while np.abs(xu - xl) > eps:

    theta_actual = (xl + xu) / 2.0
    grados_actual = np.degrees(theta_actual)
    

    print(f"Iteración {iteracion}: Ángulo estimado = {theta_actual:.4f} rad ({grados_actual:.2f}°) | Rango: [{xl:.4f}, {xu:.4f}]")
    

    if ang(x1) > ang(x2):
        xl = x2
        x2 = x1
        d = R * (xu - xl)
        x1 = xl + d
    else:
        xu = x1
        x1 = x2
        d = R * (xu - xl)
        x2 = xu - d
        
    iteracion += 1

```

```python

# Parámetros e inicialización
x_i = np.radians(45.0)
epsilon = 0.05
theta_exacto = 1.0472
iteracion = 0
error_relativo = 1.0
f_prime_val = 1.0

# Algoritmo de Newton
while abs(f_prime_val) >= epsilon and error_relativo >= epsilon:
    iteracion = iteracion + 1
    
    # Cálculo de derivadas en el punto actual x_i
    f_prime_val = 4.0 * np.cos(x_i) + 4.0 * np.cos(2.0 * x_i)
    f_dprime_val = -4.0 * np.sin(x_i) - 8.0 * np.sin(2.0 * x_i)
    
    # Iteración: x_{i+1} = x_i - f'(x_i) / f''(x_i)
    x_i_next = x_i - (f_prime_val / f_dprime_val)
    
    # Criterio de tolerancia: error relativo entre pasos
    error_relativo = abs((x_i_next - x_i) / x_i_next)
    
    # Actualización de la estimación
    x_i = x_i_next

# Evaluación de f(x_{i+1}) en la solución encontrada
theta_opt = x_i
area_max = 4.0 * np.sin(theta_opt) * (1.0 + np.cos(theta_opt))
error_absoluto = abs(theta_opt - theta_exacto)
error_relativo_porcentual = (error_absoluto / theta_exacto) * 100.0

# Resultados
print("Se encontró el valor de theta en la iteración:", iteracion)
print("Ángulo óptimo encontrado (theta*):", theta_opt, "rad")
print("Ángulo óptimo en grados:", np.degrees(theta_opt), "deg")
print("Área máxima obtenida A(theta*):", area_max)
print("Valor de f'(theta*) alcanzado:", f_prime_val)
print("Error relativo porcentual respecto al valor exacto:", error_relativo_porcentual, "%")
```
