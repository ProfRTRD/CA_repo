## Temperatura optima de emision estelar

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sigma = 5.67e-8 #W*m^-2 k^-4
T_0 = 10000 #K
hconstant = 5000 #K

def potencia(T):
    P = sigma * T**4 * np.exp(-T/T_0) * (1 - np.exp(-hconstant/T))
    return P
### Aplicar seccion Aurea
TL = 3000
TU = 50000

eps = 50 #K

R = (np.sqrt(5) - 1) / 2

d = R * (TU - TL)
T1 = TL + d
T2 = TU - d

ite = 1

while np.abs(TU - TL) > eps:

    if potencia(T1) > potencia(T2):
        TL = T2
        T2 = T1
        d = R * (TU - TL)
        T1 = TL + d
    else:
        TU = T1
        T1 = T2
        d = R * (TU - TL)
        T2 = TU - d
    
    ite = ite + 1
T_actual = (TL + TU) / 2.0
print("valor de T que maximiza la funcion:", T_actual)
    
```

```python
##Graficar P(T)
T_grafica = np.linspace(3000, 50000, 1000)
P_grafica = potencia(T_grafica)
y_max = potencia(T_actual)
plt.plot(T_grafica, P_grafica, label='Potencia P(T)')
plt.plot(T_actual, y_max, 'ro', label='Maximo')

plt.xlabel('Temperatura T (K)')
plt.ylabel('Potencia P(T)')
plt.title('Potencia vs Temperatura')
plt.legend()
plt.grid(True)
plt.show()
```
