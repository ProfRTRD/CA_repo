# Maximizar la luminosidad de un disco de acrecion simplificado

```python
import numpy as np
```

## a) Condiciones iniciales 
$$f(r_0, \theta_0) = (r_0, \pi/6)$$

```python
def luminosidad(r, theta):
    return r**2 * np.sin(theta)*(1 + np.cos(theta))* np.exp(-r)

def aurea_r(theta_fija):
    xl = 0.1  # Límite inferior
    xu = 5.0  # Límite superior
    eps = 0.05
    R = (np.sqrt(5) - 1) / 2
    
    d = R * (xu - xl)
    x1 = xl + d
    x2 = xu - d
    iteracion = 1
    
    while np.abs(xu - xl) > eps:
        r_actual = (xl + xu) / 2.0
        print(f"Iteración r {iteracion}: r estimado = {r_actual:.4f} | Rango: [{xl:.4f}, {xu:.4f}]")
        
        if luminosidad(x1, theta_fija) > luminosidad(x2, theta_fija):
            xl = x2
            x2 = x1
            d = R * (xu - xl)
            x1 = xl + d
        else:
            xu = x1
            x1 = x2
            d = R * (xu - xl)
            x2 = xu - d
            
        iteracion += 1
        
    r_opt = (xl + xu) / 2.0
    print(f"--> r óptimo encontrado: {r_opt:.4f}\n")
    return r_opt


def aurea_theta(r_fijo):
    xl = 0.0       # Límite inferior
    xu = np.pi/2.0 # Límite superior
    eps = 0.05
    R = (np.sqrt(5) - 1) / 2
    
    d = R * (xu - xl)
    x1 = xl + d
    x2 = xu - d
    iteracion = 1
    
    while np.abs(xu - xl) > eps:
        theta_actual = (xl + xu) / 2.0
        print(f"Iteración theta {iteracion}: theta estimado = {theta_actual:.4f} | Rango: [{xl:.4f}, {xu:.4f}]")
        
        
        if luminosidad(r_fijo, x1) > luminosidad(r_fijo, x2):
            xl = x2
            x2 = x1
            d = R * (xu - xl)
            x1 = xl + d
        else:
            xu = x1
            x1 = x2
            d = R * (xu - xl)
            x2 = xu - d
            
        iteracion += 1
        
    theta_opt = (xl + xu) / 2.0
    print(f"--> theta óptimo encontrado: {theta_opt:.4f} rad")
    return theta_opt


#Valor inicial
theta_inicial = np.pi / 6  

#Encontrar r optimo usando el theta inicial
r_optimo = aurea_r(theta_inicial)

#Encontrar theta optimo usando el r recién encontrado
theta_optimo = aurea_theta(r_optimo)
print(f"Punto de máxima luminosidad estimado: L({r_optimo:.4f}, {theta_optimo:.4f}) = {luminosidad(r_optimo, theta_optimo):.4f}")
```

## b) Condiciones iniciales 
$$f(r_0, \theta_0) = (3r_0, \pi/4)$$

```python
# Valores iniciales
r_inicial_b = 3.0
theta_inicial_b = np.pi / 4  

#Encontrar r optimo usando el theta inicial
r_optimo_b = aurea_r(theta_inicial_b)

#Encontrar theta optimo usando el r recién encontrado
theta_optimo_b = aurea_theta(r_optimo_b)

print(f"Punto de máxima luminosidad estimado: L({r_optimo_b:.4f}, {theta_optimo_b:.4f}) = {luminosidad(r_optimo_b, theta_optimo_b):.4f}")
```

## c) Condiciones iniciales 
$$f(r_0, \theta_0) = (0.5r_0, \pi/3)$$

```python
# Valores iniciales
r_inicial_b = 0.5
theta_inicial_b = np.pi / 3  

#Encontrar r optimo usando el theta inicial
r_optimo_b = aurea_r(theta_inicial_b)

#Encontrar theta optimo usando el r recién encontrado
theta_optimo_b = aurea_theta(r_optimo_b)

print(f"Punto de máxima luminosidad estimado: L({r_optimo_b:.4f}, {theta_optimo_b:.4f}) = {luminosidad(r_optimo_b, theta_optimo_b):.4f}")
```
