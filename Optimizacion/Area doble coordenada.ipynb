```python
def func(l, theta):
    return (18 - 2*l + 2*l * np.cos(theta)) * l *np.sin(theta)

#Usar el metodo de seccion aurea
#coordenada l
def aureal(f):
    theta = np.pi/6
    xl = 0.0
    xu = 9.0
    eps = 0.05
    R = (np.sqrt(5) - 1) / 2
    
    d = R * (xu - xl)
    x1 = xl + d
    x2 = xu - d
    
    
    iteracion = 1
    
    while np.abs(xu - xl) > eps:
    
        l_actual = (xl + xu) / 2.0
    
        print(f"Iteración {iteracion}: lado estimado = {l_actual:.4f}| Rango: [{xl:.4f}, {xu:.4f}]")
        
    
        if func(x1, theta) > func(x2, theta):
            xl = x2
            x2 = x1
            d = R * (xu - xl)
            x1 = xl + d
        else:
            xu = x1
            x1 = x2
            d = R * (xu - xl)
            x2 = xu - d
            
        iteracion += 1
    l_opt = (xl + xu) / 2.0
    print(f"l óptimo (cm*): {l_opt:.4f}")
    return l_opt
l_0 = aureal(func)

def aureatheta(f):
    xl = 0.0
    xu = np.pi/2.0
    eps = 0.05
    R = (np.sqrt(5) - 1) / 2
    
    d = R * (xu - xl)
    x1 = xl + d
    x2 = xu - d
    
    
    iteracion = 1
    
    while np.abs(xu - xl) > eps:
    
        theta_actual = (xl + xu) / 2.0
    
        print(f"Iteración {iteracion}: Ángulo estimado = {theta_actual:.4f}| Rango: [{xl:.4f}, {xu:.4f}]")
        
    
        if func(x1, l_0) > func(x2, l_0):
            xl = x2
            x2 = x1
            d = R * (xu - xl)
            x1 = xl + d
        else:
            xu = x1
            x1 = x2
            d = R * (xu - xl)
            x2 = xu - d
            
        iteracion += 1
    theta_opt = (xl + xu) / 2.0
    print(f"angulo óptimo (theta*): {theta_opt:.4f}")
    return theta_opt
aureatheta(func)  
```
