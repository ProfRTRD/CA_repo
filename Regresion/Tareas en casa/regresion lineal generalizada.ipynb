Dada la forma $y = a_1x$, y $a_1 = \frac{\sum_{i=1}^n x_i y_i}{\sum_{i=1}^n x_i^2}$, tenemos que$$E = \frac{\sum_{i=1}^n \sigma_i \varepsilon_i}{\sum_{i=1}^n \varepsilon_i^2}$$si $n = 12$, entonces$$\sum_{i=1}^{12} \varepsilon_i^2 = 1.2764 \times 10^{-3}; \quad \sum_{i=1}^{12} \sigma_i \varepsilon_i = 2.3337 \times 10^8,$$$$E = \frac{\sum_{i=1}^{12} \sigma_i \varepsilon_i}{\sum_{i=1}^{12} \varepsilon_i^2} = \frac{2.3337 \times 10^8}{1.2764 \times 10^{-3}} = 1.8284 \times 10^{11} Pa$$

# Regresion lineal generalizada
Teniendo que $y = a_0 + a_1x$ para un conjunto de datos de $n$ puntos. Y recordando que$$S_r = \sum_{i=1}^n E_i^2 = \sum_{i=1}^n (y_i - a_0 - a_1x)^2$$Y haciendo un poco de algebra como en el caso anterior, encontramos que$$a_1 = \frac{S_{xy}}{S_{xx}}; \quad a_0 = \bar{y} - a_1\bar{x}$$donde$$S_{xy} = \sum_{i=1}^n x_iy_i - n\bar{x}\bar{y}; \, S_{xx} = \sum_{i=1}^n x_i^2 - n\bar{x}^2; \, \bar{x} = \frac{\sum_{i=1}^n x_i}{n}, \, \bar{y} = \frac{\sum_{i=1}^n y_i}{n}$$

Para este ejercicio tenemos:

T = k1 + k2$\theta$
$$S_{T\theta} = \sum_{i=1}^n \theta_iT_i - n\bar{\theta}\bar{T}$$
$$S_{\theta \theta} = \sum_{i=1}^n \theta_i^2 - n\bar{\theta}^2$$

$$S_r = \sum_{i=1}^n E_i^2 = \sum_{i=1}^n (y_i - a_0 - a_1x)^2$$
$ y = T, a_0 = k1, a_1 = k2, x = theta$

$$k2 = \frac{S_{T\theta}}{S_{\theta \theta}}$$
$$k1 = bar(T) - k2\bar(\theta)$$
$$ E = T - k1 - k2 \theta$$
$$S_{total} = E^2$$

```python
theta = np.array([0.698132, 0.959931, 1.134464, 1.570796, 1.919862])
T = np.array([0.188224, 0.209138, 0.230052, 0.250965, 0.313707])

theta_bar = np.mean(theta)
T_bar = np.mean(T)
n = len(theta)

S_Ttheta = np.sum(theta * T) - n * theta_bar * T_bar
S_dbtheta = np.sum(theta**2) - n * theta_bar**2


k2 = S_Ttheta/S_dbtheta
k1 = T_bar - k2 * theta_bar

print("valor k2:", k2, "valor k1:", k1)
```
