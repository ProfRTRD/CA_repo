```python
import numpy as np
```

Siguiendo el modelo $ae^{bx}$, aplicamos el logaritmo natural a ambos lados de la ecuación$$\ln y = \ln a + bx$$
$$ y = a + exp(bx)$$
renombramos $\ln y = z$, $\ln a = a_0$ y $a_1 = b$, entonces tenemos$$z = a_0 + a_1x$$

para este ejercicio:

$\ln(L) = \ln(A e^{\lambda t})$

$\ln(L) = \ln(A) + \lambda t$

$z = \ln(L)$ 

$x = t$

$a_0 = \ln(A)$

$a_1 = \lambda$

```python
#Datos
t = np.array([0, 10, 20, 30, 40, 50]) # tiempo en días
L = np.array([1.0, 0.914, 0.836, 0.764, 0.699, 0.639]) # flujo relativo L/L_0
n = len(t)
x = t
z = np.log(L)
barx = np.mean(x)
barz = np.mean(z)
```

## a) Evaluar las constantes de regresion A y $\lambda$

$$a_1 = \frac{n \sum_{i=1}^n x_i z_i - \sum_{i=1}^n x_i \sum_{i=1}^n z_i}{n \sum_{i=1}^n x_i^2 - \left( \sum_{i=1}^n x_i \right)^2}$$$$a_0 = \bar{z} - a_1\bar{x}$$

```python
#lambda
def func_a_1(x, z, n):
    numerador = n * np.sum(x * z) - np.sum(x) * np.sum(z)
    denominador = n * np.sum(x**2) - np.sum(x)**2
    
    a_1 = numerador / denominador
    return a_1
a_1 = func_a_1(x,z,n)
lamda = a_1
#A
def func_a_0(barx, barz, a_1):
    a_0 = np.exp(barz - a_1*barx)
    return a_0

A = func_a_0(barx, barz, lamda)
print("valor de constante lambda:", lamda, "valor de  constante A:", A)
```

## b) Vida media del $^{56}\text{Co}$

$\frac{A}{2} = A e^{\lambda t}$

$0.5 = e^{\lambda t}$

$\ln(0.5) = \lambda t$

$t = \frac{\ln(0.5)}{\lambda}$

```python
vida_media = np.log(0.5)/lamda
print("la vida media del 56Co es:", vida_media)
```

## c) El flujo relativo despues de 100 dias

```python
flujo_100 = A * np.exp(lamda * 100)
print("flujo relativo despues de 100 dias:", flujo_100)
```
