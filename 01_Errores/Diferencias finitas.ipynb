```python
import numpy as np
```

```python

#Parametros
L_0 = 5.0e42
tau = 111.4
t_0 = 64
h = 2

# Función original L(t)
def L(t):
    return L_0 * np.exp(-t / tau)

#Derivada exacta L'(t)
def L_prime_exact(t):
    return -(L_0 / tau) * np.exp(-t / tau)

#dif fwrd

def diff_forward(func, x, h):
    return (func(x + h) - func(x)) / h

#dif bcwrd
def diff_backward(func, x, h):
    return (func(x) - func(x-h)) / h
#dif centr
def diff_centered(func, x, h):
    return (func(x+h) - func(x-h)) / (2*h)

#Resultados
exact_0 = L_prime_exact(0)
exact = L_prime_exact(t_0)
fwd = diff_forward(L, t_0, h)
bwd = diff_backward(L, t_0, h)
ctr = diff_centered(L, t_0, h)

print(f"Exacta en t = 0: {exact_0:.4e}")
print(f"Exacta:   {exact:.4e}")
print(f"Forward:  {fwd:.4e}")
print(f"Backward: {bwd:.4e}")
print(f"Centered: {ctr:.4e}")
```

```python
#errores relativos
err_fwd = abs((exact - fwd) / exact) * 100
err_bwd = abs((exact - bwd) / exact) * 100
err_ctr = abs((exact - ctr) / exact) * 100

print(f"Error Forward: {err_fwd:.4f}%")
print(f"Error Backward: {err_bwd:.4f}%")
print(f"Error Centered: {err_ctr:.4f}%")
```

### Derivación de las fórmulas de diferencias finitas para la segunda derivada

#### 1. Diferencia hacia adelante (Forward)
Utilizamos puntos en pasos futuros: $x_0 + h$ y $x_0 + 2h$.
$$f(x_0 + h) = f(x_0) + h f'(x_0) + \frac{h^2}{2} f''(x_0) + \mathcal{O}(h^3)$$
$$f(x_0 + 2h) = f(x_0) + 2h f'(x_0) + 2h^2 f''(x_0) + \mathcal{O}(h^3)$$

$$f''(x_0) = \frac{f(x_0 + 2h) - 2 f(x_0 + h) + f(x_0)}{h^2} + \mathcal{O}(h)$$

#### 2. Diferencia hacia atrás (Backward)
Utilizamos puntos en pasos anteriores: $x_0 - h$ y $x_0 - 2h$.
$$f(x_0 - h) = f(x_0) - h f'(x_0) + \frac{h^2}{2} f''(x_0) + \mathcal{O}(h^3)$$
$$f(x_0 - 2h) = f(x_0) - 2h f'(x_0) + 2h^2 f''(x_0) + \mathcal{O}(h^3)$$

$$f''(x_0) = \frac{f(x_0) - 2 f(x_0 - h) + f(x_0 - 2h)}{h^2} + \mathcal{O}(h)$$

#### 3. Diferencia central (Centered)
$$f(x_0 + h) = f(x_0) + h f'(x_0) + \frac{h^2}{2} f''(x_0) + \frac{h^3}{6} f'''(x_0) + \mathcal{O}(h^4)$$
$$f(x_0 - h) = f(x_0) - h f'(x_0) + \frac{h^2}{2} f''(x_0) - \frac{h^3}{6} f'''(x_0) + \mathcal{O}(h^4)$$

$$f(x_0 + h) + f(x_0 - h) = 2f(x_0) + h^2 f''(x_0) + \mathcal{O}(h^4)$$

$$f''(x_0) = \frac{f(x_0 + h) - 2f(x_0) + f(x_0 - h)}{h^2} + \mathcal{O}(h^2)$$

```python
#derivadas dobles
#exacta
def L_double_prime_exact(t):
    return (L_0 / (tau**2)) * np.exp(-t / tau)
#doble diff centered
def diff2_centered(func, x, h):
    return(func(x + h) - 2*func(x) + func(x - h))/(h**2)

```

```python
#verificaciones
exact2 = L_double_prime_exact(t_0)
ctr2 = diff2_centered(L, t_0, h)

print("\n--- Segundas Derivadas ---")
print(f"Exacta 2da: {exact2:.4e}")
print(f"Centered 2da: {ctr2:.4e}")

# Error relativo segunda derivada
err_ctr2 = abs((exact2 - ctr2) / exact2) * 100

print(f"Error Centered 2da: {err_ctr2:.4f}%")
```
