```python
import numpy as np
import math
#El error de truncar la serie
#literal a)
def seriese(x):
    return 1 + x + ((x**2)/2) + ((x**3)/math.factorial(3))
approx_val = seriese(0.5)
real_val = math.exp(0.5)
print("funcion propia:",approx_val)
print("funcion math:",real_val)

#literal b)
error_verdadero = real_val - approx_val
error_rel = abs(error_verdadero)/real_val
print("error verdadero:",error_verdadero,"error relativo verdadero:", error_rel)

#literal c)
def cota_residuo_taylor(x, n, max_derivada):
    return (max_derivada * (x**(n + 1))) / math.factorial(n + 1)

cota_superior = cota_residuo_taylor(0.5, 3, 2)

print("Cota calculada con la función R_n:", cota_superior)
#literal d)
cumple_cota = error_verdadero < cota_superior
if cumple_cota == True:
    print("cae dentro de la cota"),
else:
    print("no cae dentro de la cota")
```
