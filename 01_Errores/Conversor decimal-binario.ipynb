```python
import numpy as np
import math

#Convertor decimal binario

def entero_a_binario(n):
    # Caso especial para el 0
    if n == 0:
        return "0"
        
    resultado_binario = ""
    
    while n > 0:
        residuo = n % 2
        cociente = n // 2 
        resultado_binario = str(residuo) + resultado_binario
        n = cociente
        
    return resultado_binario

def fraccional_a_binario(f, max_bits=20):
    resultado = ""
    while f > 0 and len(resultado) < max_bits:
        f *= 2
        bit = int(f)  
        resultado += str(bit)  
        f -= bit
    return resultado
#literal a)
ex1 = entero_a_binario(18)
ex1_2 = fraccional_a_binario(0.0308)
ex2 = entero_a_binario(26)
ex2_2 = fraccional_a_binario(0.0985)
ex3 = entero_a_binario(94)
ex3_2 = fraccional_a_binario(0.0516)
print("18.0308:",ex1+"."+ex1_2)
print("26.0985:",ex2+"."+ex2_2)
print("94.0516;",ex3+"."+ex3_2)
```

```python
#IEEE 754 para float 32
#(-1)^(sign) * (1.mantisa) * 2 ^(exponente - bias)
def float32_ieee754(numero):
    # Signo
    signo = '1' if numero < 0 else '0'
    numero = abs(numero)
    
    #Separar parte entera y fraccional
    parte_entera = int(numero)
    parte_fraccional = numero - parte_entera
    
    #Conversión a binario puro
    bin_entero = entero_a_binario(parte_entera)
    bin_frac = fraccional_a_binario(parte_fraccional, max_bits=50) 
    
    #Normalización y Mantisa cruda
    if parte_entera > 0:
        #El punto se mueve len() - 1 posiciones.
        exponente_real = len(bin_entero) - 1
        mantisa_cruda = bin_entero[1:] + bin_frac
    else:
        # Si el entero es 0, buscamos el primer '1' en la fracción
        primer_uno = bin_frac.find('1')
        if primer_uno == -1:
            return signo + " " + "0"*8 + " " + "0"*23 # Caso especial: el número es 0.0
            
        exponente_real = -(primer_uno + 1)
        # La mantisa es todo lo que sigue después del primer '1'
        mantisa_cruda = bin_frac[primer_uno + 1:]
        
    #Exponente (Bias 127)
    exponente_bias = exponente_real + 127
    bin_exp = entero_a_binario(exponente_bias)
    bin_exp = bin_exp.zfill(8) #Asegura que tenga 8 bits rellenando con ceros a la izquierda
    
    #Mantisa a 23 bits
    mantisa_23 = mantisa_cruda[:23].ljust(23, '0')
    
    #Ensamblaje final
    return f"{signo} {bin_exp} {mantisa_23}"

numeros = [18.0308, 26.0985, 94.0516]
for n in numeros:
    print(f"{n} -> {float32_ieee754(n)}")
```
