```python
import numpy as np
import math
```

```python
#Representacion en series de la suma de taylor para Asin(t*omega)
#Parametros
A = 0.1
omega = 3.105
t = 1
deltat = 0.2
#representacion en taylor
def TaylorSin(k_max):
    suma = 0
    x = omega * t

    for k in range(k_max + 1):
        numerador = ((-1)**k) * (x**(2*k))
        denominador = math.factorial(1 + 2 * k)
        suma += numerador / denominador
    return A * x * suma
#Resultados
suma6 = TaylorSin(6)
suma7 = TaylorSin(7)
print("suma truncada en k = 6:",suma6)
print("suma truncada en k = 7:",suma7)
#Error approx
e_approx = abs(suma7 - suma6)/ suma7
porcentaje = e_approx * 100

print("error aproximado:", e_approx)

#Criterio de scarborough
m = 2 - math.log10(2 * porcentaje)
print("Valor de m:", m)
```

```python
#En que valor de k se obtienen 7 cifras significativas
#m = 7 cte
#Criterio de scarborough
es = 0.5 * (10**(2-7))
k = 0
x = omega * t
suma_anterior = 0
suma_actual = 0
ea = 100
while ea > es:
    numerador = ((-1)**k) * (x**(2*k))
    denominador = math.factorial(1 + 2 * k)    
    suma = numerador / denominador

    suma_actual = suma_anterior + suma
    valor_aproximado = A * x * suma_actual

    if k > 0:
        valor_anterior = A * x * suma_anterior
        ea = abs((valor_aproximado - valor_anterior) / valor_aproximado) * 100
        
    if ea > es:
        suma_anterior = suma_actual
        k += 1
print(f"Se requieren {k} iteraciones.")
print(f"El valor aproximado es: {valor_aproximado}")

```
