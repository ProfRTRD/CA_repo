# Liberacion de energia de una fulguracion solar

```python
import numpy as np
```

```python
#definir funcion
#funcion: dE/dt = -alfa * E^n
def dE_dt(alfa, E):
    alfa = 0.5
    n = 1.5
    return -alfa * E**n
#Construir rk4 con h = 0.01, E(0) = E0 = 1, alfa = 0.5, n = 1.5

def RK4(dE_dt, E0, t0, t_max, h):
    E = E0
    t_actual = t0

    E_return = [E]
    t_return = [t_actual]
    

    while t_actual < t_max:

        k1x = h * dE_dt(t_actual, E)
        
        k2x = h * dE_dt(t_actual + h/2, E + k1x/2)
        
        k3x = h * dE_dt(t_actual + h/2, E + k2x/2)
        
        k4x = h * dE_dt(t_actual + h, E + k3x)
        

        E += (1/6) * (k1x + 2*k2x + 2*k3x + k4x)
        
        t_actual += h

        E_return.append(E)
        t_return.append(t_actual)
    return t_return, E_return

#Intervalo de tiempo [0, 10]
t_res, E_res = RK4(dE_dt, E0 = 1, t0 = 0, t_max = 10, h = 0.01)

print('en el momento:', t_res[-1], 'el valor de la energia es:',  E_res[-1])
```
