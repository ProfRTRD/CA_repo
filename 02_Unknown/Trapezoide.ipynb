```python
import numpy as np

# Parámetros del problema
m0 = 140000
q = 2100
u = 2000
g = 9.81
a = 8    
b = 30
valor_verdadero = 11057.16

def v(t):
    return u * np.log(m0 / (m0 - q * t)) - g * t
def trapezoide(a, b, n_max):
    ###-----Encabezado tabla-----###
    print(f"{'n':<6}{'Estimacion x':<15}{'Error Abs':<15}{'Error rel (%)':<15}")
    print("-" * 55)
    #pasos
    for n in range(1, n_max +1):
        h = (b-a)/n
        sumando = 0
    #Regla del trapezoide
    #I = #(b-a)/(2*n) * [f(a) + 2*sum(f(x_i)) + f(b)  
        for i in range(1, n):
            x_i = a + i*h
            sumando = sumando + v(x_i)
            
        Estimacion = (b-a)/(2*n) * (v(a) + 2 * sumando + v(b))
        #Error absoluto:
        error_absoluto = np.abs(valor_verdadero - Estimacion)
        #Error relativo:
        error_relativo = (error_absoluto / np.abs(valor_verdadero)) * 100
        #Imprimir resultados en la tabla
        print(f"{n:<6}{Estimacion:<15.4f}{error_absoluto:<15.4f}{error_relativo:<15.4f}")
#Ejecucion        
trapezoide(a, b, n_max=8)
```
