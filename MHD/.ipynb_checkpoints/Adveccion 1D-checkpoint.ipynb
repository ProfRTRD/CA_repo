```python
#Caso base
import numpy as np
import matplotlib.pyplot as plt

#Parametros
N = 200
L = 1.0
v0 = 1.0
dx = L / N
dt = 0.4 * dx / v0 #parametro CFL
x = np.linspace(0, L, N, endpoint =False)
B = np.sin(2 * np.pi * x / L)
for _ in range (200):
    backward = B - np.roll(B, 1)
    B += -(v0 * dt /dx) * backward
plt.figure()
plt.plot(x, B, color='skyblue')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

#Parametros
N = 150
L = 3.0
v0 = 10.0
dx = L / N
dt = 1.8 * dx / v0 #parametro CFL
x = np.linspace(0, L, N, endpoint =False)
B = np.sin(2 * np.pi * x / L)
for _ in range (200):
    backward = B - np.roll(B, 1)
    B += -(v0 * dt /dx) * backward
plt.figure()
plt.plot(x, B, color='skyblue')
plt.show()
```
